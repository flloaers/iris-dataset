**Principal Component Analysis** (PCA) is the art of projecting data onto a lower-dimensional hyperplane — not just *any* hyperplane, but the one that **maximises** the variance of the projected data.

### **Solution**:
The core idea of PCA is to find a lower-dimensional hyperplane (or subspace) that maximises variance upon data projection, thus capturing the essential information within fewer dimensions.


# Standardise the data

**Principal Component Analysis is sensitive to the scale of the features.** Remember: PCA aims to find the directions of maximum variance in the data. If features have different scales, PCA be dominated by features with larger variances simply because of their scale, not because they are inherently more informative. This would lead to biased results. To address this issue, we standardise the data by transforming the features to have a mean of zero and a standard deviation of one (unit variance). 

The `fit` part of the `fit_transform()` method from `sklearn.preprocessing.StandardScaler` calculates the mean and the variance of each feature, while the `transform` part transforms the features according to 
```math
x \rightarrow \frac{x - \bar{x}}{\sigma},
```
where $\bar{x}$ is the sample mean of a feature and $\sigma$ is its standard deviation, i.e., the square root of its variance.

Alternatively, this step can of course be implemented by using the `numpy.mean()` and `numpy.std()` functions. In this case, we need to specify the axis along which the mean and standard deviation are to be computed. 

````python
# Prepare data for standardisation. 

# Create feature matrix (150 samples x 4 features).
X = iris_df.drop(columns='species').to_numpy()
# Alternatively: 
# X = iris_df.iloc[:, 0:4].to_numpy()

# Create target vector (150 samples, species categories).
Y = iris_df.species.to_numpy()

# Standardise the features by removing the mean & scaling to unit variance.
X = StandardScaler().fit_transform(X)

# Alternative using numpy functions 
# X = (X - np.mean(X, axis=0)) / np.std(X, axis=0)

# Let's check. 
print(f"Means: {np.mean(X, axis=0)}.")
print(f"Standard deviations: {np.std(X, axis=0)}.")
````

# Build the covariance matrix

**Covariance** tells us both the direction and the strength of the linear relationship between two variables. A positive covariance indicates that the variables tend to increase or decrease in tandem, while a negative covariance indicates that they move in opposite directions. The magnitude of the covariance reflects the strength of the relationship. 

The covariance between two variables $X_i$ and $X_j$, denoted as $\sigma_{ij}$ or $\text{cov}(X_i,X_j)$, is defined as
```math
\sigma_{ij} = \text{cov}(X_i, X_j) = \frac{1}{N-1} \sum_{k=1}^{N}(X_{i,k} - \bar{X}_i)(X_{j,k} - \bar{X}_j),
```
where $\bar{X}_i$ (resp. $\bar{X}_j$) is the sample mean of $X_i$ (resp. the sample mean of $X_j$).

When we have multiple variables, the pairwise covariances can be organised into a square matrix $\mathbf{K}$, called the **covariance matrix**, whose element $K_{ij}$ is the covariance $\sigma_{ij}$. This matrix
* is symmetric, since $\sigma_{ij} = \sigma_{ji}$, 
* is positive semi-definite, i.e., $\mathbf{A}^{\mathsf{T}} \mathbf{K} \mathbf{A} \geq 0$ for all $\mathbf{A} \in \mathbb{R}^n$, 
* contains the variances, i.e., the covariance of each element with itself, $\sigma_{ii}$, on its main diagonal. 

When fed a 2D-array, the `numpy.cov()` function, by default, interprets the *rows* as *features* (or variables) and the *columns* as *observations* (or samples) of said features. In other words, since our feature matrix `X` has features as columns and observations as rows (which is the typical convention in machine learning), we should then pass the *transpose* `X.T` of the feature matrix to `numpy.cov()`. Alternatively, we can set the optional `rowvar` keyword parameter to `False`. 


````python
covariance_matrix = np.cov(X.T)
print("Covariance matrix: \n", covariance_matrix)
````

# Find the eigensystem of the covariance matrix

### Principal components 
The principal components of a dataset are new variables that are built as *linear combinations* of the original features in such a way that
* the **first** principal component accounts for the largest possible variance in the data; 
* the $i\!$ th principal component accounts for the $i\!$ th highest variance, while being orthogonal (i.e., uncorrelated) to the $(i-1)$ preceeding components).

In other words, the principal components provide a new orthogonal basis for the data. Keeping only part of this coordinate system makes it possible to reduce the dimensionality of the system while preserving most of the information.

### Eigenvectors & eigenvalues
One way to determine these principal components is to compute the eigensystem of the covariance matrix $\mathbf{K}$. The eigenvectors of $\mathbf{K}$ represent the **directions** of the principal components of the dataset, while each eigenvalue accounts for the variance explained by its associated eigenvector. By computing the eigensystem of $\mathbf{K}$ and ordering the eigenvectors by their corresponding descending eigenvalues, we can access the *ordered* principal components of the dataset. 

In the case of the Iris dataset, the covariance matrix is of size $4 \times 4$. It therefore has four eigenvectors, $\mathbf{w}_1, \mathbf{w}_2, \mathbf{w}_3, \mathbf{w}_4$, and four eigenvalues, $\lambda_1, \lambda_2, \lambda_3, \lambda_4$, such that 
```math 
\mathbf{K} \mathbf{w}_i = \lambda_k \mathbf{w}_{i} \quad \text{for } i \in \lbrace1, 2, 3, 4\rbrace.
```
Defining $\mathbf{W}$ as the square matrix whose $i\!$ th column is the eigenvector $\mathbf{w}_i$, we can also write the so-called *eigendecomposition* of $\mathbf{K}$ as
```math 
\mathbf{K} = \mathbf{W} \boldsymbol{\Lambda} \mathbf{W}^{-1}, 
```
where $\boldsymbol{\Lambda}$ is the diagonal matrix whose diagonal elements are the corresponding eigenvalues of $\mathbf{K}$, that is,
```math
\boldsymbol{\Lambda} = \begin{pmatrix} \lambda_1 & 0 & 0 & 0 \\ 
0 & \lambda_2 & 0 & 0 \\ 0 & 0 & \lambda_3 & 0 \\ 0 & 0 & 0 & \lambda_4 \end{pmatrix}.
```


````python
eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)
print(f'Eigenvectors: \n {eigenvectors}. \n')
print(f'Eigenvalues: \n {eigenvalues}.')
````

Since the covariance matrix $\mathbf{K}$ is a real symmetric matrix, we could also use the `numpy.linalg.eigh()` function, which works for complex Hermitian matrices. Note that the `numpy.linalg.eig()` function does not necessarily return the eigenvalues ordered, while the `numpy.linalg.eigh()` function return them in *ascending* order. 

In the above case, the returned eigenvalues *happen* to be conveniently sorted in descending order. What if they are not? To put both the eigenvalues and the corresponding eigenvectors in descending order, the `numpy.argsort()` function comes in handy.

````python
# In case sorting is necessary
idx = np.argsort(eigenvalues) # returns the array of indices that would sort eigenvalues in ascending order
idx = idx[::-1] # reverses the order of the indices
eigenvalues = eigenvalues[idx] # sorts the eigenvalues
eigenvectors = eigenvectors[:,idx] # sorts the eigenvectors accordingly
````

# Explain the variance

We are left with a key question: how many principal components should we keep, and how can we quantitatively justify our choice?

Here it is useful to compute and plot the **[cumulative] explained variance** as a function of the principal component number. The explained variance associated with the $i\!$ th principal component quantifies how much of the total variance in the original data is captured by that particular principal component. In other words, if $\lambda_i$ is the eigenvalue associated with the $i\!$ th principal component, its explained variance is given by 
```math
\text{EV}(i) = \frac{\lambda_i}{\sum_{i=1}^{n} \lambda_i},
```
where $n$ is the total number of principal components, i.e., the number of original variables. As will be the case in the code below, explained variances are often expressed as percentages. Naturally, the more principal components we consider, the more variance we capture. By summing the explained variance of the first $k$ principal components, we obtain the **cumulative** explained variance, which quantifies the total variance captured up to that point,  
```math
\text{CEV}(k) = \sum_{i=1}^{k} \text{EV}(i).
```

````python
# Calculate the explained variance associated with each of the principal components
explained_variances = [(v/sum(eigenvalues))*100 for v in eigenvalues]
# Calculate the cumulative explained variances 
cumulative_explained_variances = np.cumsum(explained_variances)

# Gather the values in a DataFrame
explain_data = pd.DataFrame({
    'Principal Component': range(1, len(eigenvalues) + 1),
    'Individual Explained Variance (in %)': explained_variances,
    'Cumulative Explained Variance (in %)': cumulative_explained_variances
})

# Display
with pd.option_context('expand_frame_repr', False):
    print(explain_data)

fig, ax = plt.subplots()

d = 4 # number of original variables/features
plt.bar(range(1, d+1), explained_variances, align='center',
        label='Individual Explained Variance', color = 'khaki')

plt.plot(range(1, d+1), cumulative_explained_variances, marker="o",
         label='Cumulative Explained Variance', color='#00B9FF')

## add cumulative variance annotation
for x, y in zip(range(1, d+1), cumulative_explained_variances):
    ax.annotate(f"{y:.1f}%", (x - 0.2, y + 1.4))

## add label, title, ticks
plt.xlabel('Principal Component Number')
plt.ylabel('Explained Variance (%)')
plt.title('Cumulative Explained Variance')
plt.xticks(range(1, d+1))

## legend location
plt.legend(loc='best')
plt.show()

````

# Check your elbow

A common strategy to decide how many principal components to keep is to use the  **elbow method**.

By plotting the cumulative explained variance as a function of the number of components, as we did above, we often observe a curve that rises steeply at first and then gradually levels off.
The elbow of the curve — the point where the rate of increase noticeably slows down — points to a good trade-off between reducing dimensionality and preserving most of the variance.

- Before the elbow, each additional principal component captures a significant portion of the remaining variance.
- After the elbow, adding more components yields only marginal gains.

Selecting the number of components corresponding to this method allows us to simplify the dataset without sacrificing too much information.

In our case, keeping just **two components** is enough to explain over 95% of the total variance — a strong indicator that dimensionality can be safely reduced. 

# Project the data

Our original feature space is $d$-dimensional, where $d = 4$ in the Iris dataset. The eigenvectors corresponding to the top $k$ principal components form a $d \times k$ matrix $\mathbf{P}$ that we can use to *project* the original data onto a new, k-dimensional subspace. 

Projection is achieved by multiplying the original $n \times d$ feature matrix by $\mathbf{P}$. In the Iris dataset, we have $n = 150$. Keeping the top 2 principal components thus yields a $150 \times 2$ matrix.

````python
# Define the projection matrix P
P = eigenvectors[:, :2]
print('Projection matrix: \n', P)

# Project the data on the two-dimensional feature space
X_PCA2 = X.dot(P)

# Plot the transformed data
for species in iris_df['species'].unique():
    sns.scatterplot(x = X_PCA2[Y==species, 0], 
                    y = X_PCA2[Y==species, 1], 
                    label=species)
    
#Customise the labels and the title
plt.xlabel('Principal Component 1')  
plt.ylabel('Principal Component 2') 
plt.title('PCA of the Iris Dataset: from 4D to 2D')
plt.show() 
````

## **Observations**

Even though we have reduced our original data **by half**, from a $150 \times 4$ matrix to a $150 \times 2$ matrix, the transformed data retains a clear structure, allowing for good separation between the three Iris species, which is promising for tasks such as clustering or classification. 

This further confirms that most of the variance in the original data is captured by the first two principal components. This dimensionality reduction not only simplifies the data but also provides a more efficient representation while preserving the key features needed for distinguishing the species.

### **Takeaways**:
- PCA reduces the dimensionality of the dataset while preserving class structure.
- The first two components capture the majority of variance.
- Clear separation supports the use of PCA for clustering or classification.

# Explore other dimensions

We can also explore what the dataset looks like when we keep three components instead. In this case, the transformed data lives in 3D space, which we can visualize with a 3D scatter plot.

````python
from mpl_toolkits.mplot3d import Axes3D  # for 3D plotting

# Define the projection matrix P
P = eigenvectors[:, :3]
print('Projection matrix: \n', P)

# Project the data on the three-dimensional feature space
X_PCA3 = X.dot(P)


# Build a 3D scatter plot
palette = sns.color_palette("husl", 8)

fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection="3d", elev=-150, azim=110)

for i, species in enumerate(iris_df['species'].unique()):
    ax.scatter(
        X_PCA3[Y==species, 0],
        X_PCA3[Y==species, 1],
        X_PCA3[Y==species, 2],
        color=palette[i],
        label=species
    )

# Customise the labels and legend
ax.set_xlabel("Principal Component 1", labelpad=10)
ax.set_ylabel("Principal Component 2", labelpad=10)
ax.set_zlabel("Principal Component 3", labelpad=10)
ax.tick_params(labelsize=10)
ax.legend()
plt.title('PCA of the Iris Dataset: from 4D to 3D')
plt.tight_layout()
plt.show()
````

What if we keep only one component? We then end up with a **line** that still does an impressive job of separating the Iris species!  This should not be too surprising since the first principal component alone accounts for **72.8%** of the total variance. 

````python
# Define the projection matrix P
P = eigenvectors[:, :1]
print('Projection matrix: \n', P)

# Project the data on the three-dimensional feature space
X_PCA1 = X.dot(P)
    
sns.stripplot(x=X_PCA1[:, 0],  
              hue=iris_df['species'],  
              dodge=False,              # keeps all points on the same line
              jitter=False,             # keeps points aligned on the x-axis
              )  

#Customise the labels and the title
plt.xlabel('Principal Component 1')  
plt.title('PCA of the Iris Dataset from 4D to 1D')
plt.show() 
````

# Take a shortcut

Although we have *manually* computed the principal components to better understand the inner workings of PCA, in practice, we often rely on tools like the ``PCA`` class from the ``sklearn.decomposition`` module for convenience and efficiency. Under the hood, this implementation uses the ``scipy.linalg`` version of the **Singular Value Decomposition** (SVD) algorithm — an alternative approach to derive the principal components. 

Let us now walk through how to apply this built-in ``PCA`` class. 



````python
from sklearn.decomposition import PCA 

# Initialise the PCA model to reduce the datase to two components
pca = PCA(n_components=2)

# Fit the PCA model and transform the data
# (PCA expects a NumPy array as input)
X_PCA_sklearn = pca.fit_transform(X)

# Plot the transformed data 
for species in iris_df['species'].unique():
    sns.scatterplot(x = X_PCA_sklearn[Y==species, 0], 
                    y = X_PCA_sklearn[Y==species, 1], 
                    label=species)
    
#Customise the labels and the title
plt.xlabel('Principal Component 1')  
plt.ylabel('Principal Component 2') 
plt.title('sklearn-PCA of the Iris Dataset: from 4D to 2D')
plt.show() 
````
